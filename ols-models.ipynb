# Mini Project 2- Exploring AI Use by Demographic Features

## Using Dataset: Pew Research ATP W152, August 2024

```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
```

Include cleaned dataset, prepared in `atp-w152.ipynb`

```python
data = pd.read_csv('data/August 2024/atp-w152-clean.csv')
```

# OLS Regression Models predicting `USEAI_W152_R` likert

Reverse-coded in `atp-w152.ipynb` so higher = more frequent AI use:

- 5	Almost constantly
- 4	Several times a day
- 3	About once a day
- 2	Several times a week
- 1	Less often

Positive coefficients therefore mean *more* frequent AI use; negative coefficients mean *less* frequent.

## OLS Model 1: Income Only

```python
# USEAI_W152_R (Likert 1-5, reverse-coded: higher = more frequent AI use) regressed on household income only
model1 = smf.ols('USEAI_W152_R ~ C(F_INC_SDT1)', data=data).fit()
print(model1.summary())
```

## OLS Model 2: Education Only

```python
# USEAI_W152_R regressed on education only
model2 = smf.ols('USEAI_W152_R ~ C(F_EDUCCAT)', data=data).fit()
print(model2.summary())
```

## OLS Model 3: Income + Education

```python
# Additive model: income and education together
model3 = smf.ols('USEAI_W152_R ~ C(F_INC_SDT1) + C(F_EDUCCAT)', data=data).fit()
print(model3.summary())
```

## OLS Model 4: Income * Education

```python
# Interaction model: '*' expands to both main effects + the interaction term
model4 = smf.ols('USEAI_W152_R ~ F_INC_SDT1 * F_EDUCCAT', data=data).fit()
print(model4.summary())
```

## OLS Model 5: Income + Education + Age + Gender + Race

```python
# Full model: income/education/age as continuous ordinal scales;
# gender and race wrapped in C() since they are nominal categories
model5 = smf.ols(
    'USEAI_W152_R ~ C(F_INC_SDT1) + C(F_EDUCCAT) + C(F_AGECAT) + C(F_GENDER) + C(F_RACECMB)',
    data=data
).fit()
print(model5.summary())
```

# Summary Table: SE, 95% Confidence Intervals, and p-values across all models

One row per coefficient per model (long/tidy format). Columns: `model`, `term`, standard error (`SE`), the 95% confidence interval bounds (`CI_low`, `CI_high`), and the p-value (`p`). The coefficient (`coef`) is included for context.

```python
# Combine SE, 95% CI, and p-values from every model into one long-format table.
# One row per (model, coefficient).
models = {
    'M1: Income':    model1,
    'M2: Education': model2,
    'M3: Inc+Edu':   model3,
    'M4: Inc*Edu':   model4,
    'M5: Full':      model5,
}

rows = []
for name, res in models.items():
    ci = res.conf_int()
    rows.append(pd.DataFrame({
        'model':   name,
        'term':    res.params.index,
        'coef':    res.params.values,
        'SE':      res.bse.values,
        'CI_low':  ci[0].values,
        'CI_high': ci[1].values,
        'p':       res.pvalues.values,
    }))

summary_table = pd.concat(rows, ignore_index=True).round(3)
summary_table
```

## Model fit statistics

R-squared, adjusted R-squared, and AIC are model-level summaries (one value per model), so they go in their own table. Number of observations (`N`) is included since it varies across models due to listwise deletion of missing values.

```python
# Model-level fit statistics: R-squared, adjusted R-squared, AIC (one row per model).
fit_stats = pd.DataFrame({
    'model':         list(models.keys()),
    'R-squared':     [res.rsquared for res in models.values()],
    'Adj R-squared': [res.rsquared_adj for res in models.values()],
    'AIC':           [res.aic for res in models.values()],
    'N':             [int(res.nobs) for res in models.values()],
}).round({'R-squared': 3, 'Adj R-squared': 3, 'AIC': 1})

fit_stats
```
