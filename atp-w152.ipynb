## Subset for Regression Models
- `QKEY` is the unique identifier (id) per row
- Including predictor and outcome variables planned for logit and ols models
- Outcome variables: 
    - `CHATUSE_W152` is binary (coded 1, 2) for logit
    - `USEAI_W152` is likert for ols

```python
predictor_cols = ['F_INC_SDT1', 'F_EDUCCAT2', 'F_AGECAT', 'F_GENDER', 'F_RACECMB', 'F_RACETHNMOD']
outcome_cols = ['CHATUSE_W152', 'USEAI_W152']
include_cols = ['QKEY'] + predictor_cols + outcome_cols
data = data[include_cols]
data.head(20)
```

## Recode Variables

```python
# 99 = "Refused" code; recode as missing
recode_99_cols = predictor_cols + outcome_cols
for col in recode_99_cols:
    data[col] = data[col].replace(99, np.nan)

# Remap Binary Outcome Variable
data['CHATUSE_W152'] = data['CHATUSE_W152'].replace({2: 0})  # 1 = Yes, have used, 0 = No, have not used

display(data.describe())
display(data.head(20))
```

```python
# Save cleaned/recoded data for regression analysis
data.to_csv('data/August 2024/atp-w152-clean.csv', index=False)
```

## Logistic Regression

```python
data = data.dropna()
data.describe()
```

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf

np.random.seed(42)

# Gender_2 Female
# Gender_3 In some other way
gender_dummies = pd.get_dummies(data['F_GENDER'], prefix='Gender', drop_first=True, dtype=int)
race_dummies = pd.get_dummies(data['F_RACECMB'].astype(str), prefix='Race', drop_first=True, dtype=int)

X = pd.concat([data[['F_INC_SDT1','F_EDUCCAT2','F_AGECAT']], gender_dummies, race_dummies], axis=1)
y = data["CHATUSE_W152"]

X = sm.add_constant(X)

# Renaming columns for interpretability
rename_dict = {
    "const": "Intercept",
    "F_INC_SDT1": "Income",
    "F_AGECAT": "Age",
    "F_EDUCCAT2": "Education",
    "Gender_2.0": "Gender: Female",
    "Gender_3.0": "Gender: In some other way",
    "Race_2.0": "Race: Black or African-American",
    "Race_3.0": "Race: Asian or Asian-American",
    "Race_4.0": "Race: Mixed Race",
    "Race_5.0": "Race: Other"  
}

X = X.rename(columns=rename_dict)

logreg = sm.Logit(y, X)
logreg_results = logreg.fit()
```

```python
print(logreg_results.summary()) # the regular table is not very intepretable because of log odds
```

```python
df_odds = logreg_results.conf_int()
df_odds["Odds Ratio"] = logreg_results.params
df_odds.columns = ["CI lower", "CI Upper", "Odds Ratio"]

# converting from log odds to odds ratios for intepretabiliy
df_odds = np.exp(df_odds)

# percent change
df_odds["% Change"] = (df_odds["Odds Ratio"] - 1) * 100

# CI effect
conditions = [
    (df_odds["CI lower"] > 1.0) & (df_odds["CI Upper"] > 1.0),  # Entirely above 1
    (df_odds["CI lower"] < 1.0) & (df_odds["CI Upper"] < 1.0),  # Entirely below 1
]
choices = ["Positive Effect", "Negative Effect"]
df_odds["CI Effect"] = np.select(conditions, choices, default="No Effect")

# P-value
df_odds["P-value"] = logreg_results.pvalues
df_odds["Significant"] = df_odds["P-value"] < 0.05

df_odds = df_odds[["Odds Ratio","% Change","CI lower","CI Upper","CI Effect","P-value","Significant"]]

df_odds
```

```python
import matplotlib.pyplot as plt

plot_data = df_odds.drop(index="Intercept", errors="ignore").sort_values(by="Odds Ratio")

plt.figure(figsize=(8, 5))

# Plot the odds ratios as dots and the CIs as horizontal error bars
plt.errorbar(x=plot_data['Odds Ratio'], y=plot_data.index, 
             xerr=[plot_data['Odds Ratio'] - plot_data['CI lower'], plot_data['CI Upper'] - plot_data['Odds Ratio']],
             fmt='o', color='black', ecolor='teal', capsize=5, markersize=7)

# Add a vertical anchor line at 1.0 (The line of no effect)
plt.axvline(x=1.0, color='red', linestyle='--', linewidth=1.5)

plt.xlabel('Odds Ratio (95% Confidence Interval)')
plt.title('Demographic Predictors of AI Usage')
plt.grid(axis='x', linestyle=':', alpha=0.6)
plt.show()
```

## Interpretation

### Income
As CI bounds are above 1.0, income has a positive effect on AI usage. The result is significant (p < 0.05). With one unit increase in income, the odds of using AI increase by 7.48%. 

### Education
As CI bounds are above 1.0, education has a positive effect on AI usage. The result is significant (p < 0.05). With one unit increase in education, the odds of using AI increase by 27.20%. 

### Age
As CI bounds are below 1.0, age has a negative effect on AI usage. The result is significant (p < 0.05). For each one bracket increase in the age category (moving up from 18–29, to 30–49, to 50–64, and 65+), the odds of using AI decrease by 48.75%. 

### Gender: Female [baseline Male]
As CI bounds are below 1.0, female gender has a negative effect on AI usage. The result is significant (p < 0.05). For females the odds of using AI decrease by 30.39% compared to males. 

### Gender: In some other way [baseline Male]
As CI bounds include 1.0, there is no significant effect of "in some other way" gender on AI usage. The result is not significant (p > 0.05). 

### Race: Black or African-American [baseline White]
As CI bounds include 1.0, there is no significant effect of Black or African-American Race on AI usage. The result is not significant (p > 0.05). 

### Race: Asian or Asian-American [baseline White]
As CI bounds are above 1.0, Asian or Asian-American Race has a positive effect on AI usage. For Asian or Asian-American Race the odds of using AI increase by 63.05% compared to White. 

### Race: Mixed Race [baseline White]
As CI bounds include 1.0, there is no significant effect of Mixed Race on AI usage. The result is not significant (p > 0.05). 

### Race: Other [baseline White]
As CI bounds include 1.0, there is no significant effect of Black or Other Race on AI usage. The result is not significant (p > 0.05).
