## Subset for Regression Models
- `QKEY` is the unique identifier (id) per row
- Including predictor and outcome variables planned for logit and ols models
- Outcome variables: 
    - `CHATUSE_W152` is binary (coded 1, 2) for logit — will be remapped to 1/0
    - `USEAI_W152` is likert for ols — will be reverse-coded into `USEAI_W152_R` so higher = more frequent AI use (departs from Pew's original 1 = almost constantly … 5 = less often)

```python
predictor_cols = ['F_INC_SDT1', 'F_EDUCCAT', 'F_EDUCCAT2', 'F_AGECAT', 'F_GENDER', 'F_RACECMB', 'F_RACETHNMOD']
outcome_cols = ['CHATUSE_W152', 'USEAI_W152']
include_cols = ['QKEY'] + predictor_cols + outcome_cols
data = data[include_cols]
data.head(20)
```

```python
# Value labels from the codebook — note the two variables use DIFFERENT codings
# (e.g. code 1 means "College graduate+" in F_EDUCCAT but "Less than high school" in F_EDUCCAT2)
# Note: Comparing the original F_EDUCCAT coding (before recoding below); for analysis will use F_EDUCCAT_R for ease of interpretation
educcat_labels = {
    1: 'College graduate+',
    2: 'Some College',
    3: 'H.S. graduate or less',
    99: 'Refused',
}
educcat2_labels = {
    1: 'Less than high school',
    2: 'High school graduate',
    3: 'Some college, no degree',
    4: 'Associate degree',
    5: 'College graduate/some post grad',
    6: 'Postgraduate',
    99: 'Refused',
}

# F_EDUCCAT2 (6-category detail) collapses into F_EDUCCAT (3-category).
# Crosstab shows how the detailed categories roll up + the marginal counts.
edu_compare = pd.crosstab(
    data['F_EDUCCAT2'].map(educcat2_labels),
    data['F_EDUCCAT'].map(educcat_labels),
    margins=True, margins_name='Total',
)
edu_compare
```

## Recode Variables

```python
# 99 = "Refused" code; recode as missing
recode_99_cols = predictor_cols + outcome_cols
for col in recode_99_cols:
    data[col] = data[col].replace(99, np.nan)

# Remap Binary Outcome Variable
data['CHATUSE_W152'] = data['CHATUSE_W152'].replace({2: 0})  # 1 = Yes, have used, 0 = No, have not used

# Reverse-code USEAI_W152 into USEAI_W152_R so higher = more frequent AI use
# Original Pew coding: 1 = Almost constantly ... 5 = Less often
# Recoded (USEAI_W152_R): 5 = Almost constantly ... 1 = Less often
data['USEAI_W152_R'] = 6 - data['USEAI_W152']
data = data.drop(columns=['USEAI_W152'])
outcome_cols = ['CHATUSE_W152', 'USEAI_W152_R']

# Reverse-code F_EDUCCAT into F_EDUCCAT_R so higher = more education
# Original Pew coding: 1 = College Graduate+, 2 = Some College, 3 = H.S. graduate or less
# Recoded (F_EDUCCAT_R): 1 = H.S. graduate or less, 2 = Some College, 3 = College Graduate+
data['F_EDUCCAT_R'] = data['F_EDUCCAT'].map({
    1: 3,
    2: 2,
    3: 1,
})

# Use the reverse-coded education variable as the predictor going forward
predictor_cols = ['F_EDUCCAT_R' if col == 'F_EDUCCAT' else col for col in predictor_cols]
data = data.drop(columns=['F_EDUCCAT'])  # drop the original so it isn't carried into the saved CSV

display(data.describe())
display(data.head(20))
```

## Compare Outcome Variables

`CHATUSE_W152` (binary: 1 = used ChatGPT, 0 = not) and `USEAI_W152_R` (Likert 1–5, reverse-coded above so 5 = almost constantly and 1 = less often) both capture AI use, so we expect them to be positively related.

- Spearman is the appropriate test since `USEAI_W152_R` is ordinal and `CHATUSE_W152` is binary.
- Pearson is reported alongside (equivalent to point-biserial when one variable is dichotomous).
- With only two variables, the correlation matrix is essentially the single coefficient repeated—included for completeness, but the test above is what matters.

```python
from scipy import stats

# Use only rows with both outcomes observed
pair = data[outcome_cols].dropna()

spearman = stats.spearmanr(pair['CHATUSE_W152'], pair['USEAI_W152_R'])
pearson = stats.pearsonr(pair['CHATUSE_W152'], pair['USEAI_W152_R'])

print(f"N (complete pairs): {len(pair)}")
print(f"Spearman rho = {spearman.statistic:+.3f}, p = {spearman.pvalue:.3g}")
print(f"Pearson  r   = {pearson.statistic:+.3f}, p = {pearson.pvalue:.3g}")

# 2x2 correlation matrix (Pearson)
data[outcome_cols].corr()
```

Interpretation: after the reverse coding, both outcomes point the same way — higher = more AI engagement (`CHATUSE_W152`: 1 = used / 0 = not; `USEAI_W152_R`: 5 = almost constantly … 1 = less often). A **positive** correlation therefore means chatbot users report more frequent AI use overall, which is the substantively expected direction. The magnitude tells you how tightly the two questions track each other.
